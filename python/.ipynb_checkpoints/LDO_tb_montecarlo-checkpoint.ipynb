```python
# =========================================================
# CELDA 2: EJECUCIÓN DE NGSPICE CON CONTROL DE COMPATIBILIDAD
# =========================================================
import subprocess, os, shutil, time
from PyLTSpice import RawRead

# 1. Definir rutas de netlist y directorio de salida
netlist_path = "/headless/.xschem/simulations/ldo_tb_montecarlo.spice"
raw_path = "/foss/designs/LDO-sky130/xschem/spice_raw_log/raw/"
os.makedirs(raw_path, exist_ok=True)

# Copiar .spiceinit a las rutas requeridas sin fallar por permisos
spinit_src = "/foss/pdks/sky130A/libs.tech/ngspice/spinit"
if os.path.exists(spinit_src):
    for target in [os.path.join(raw_path, ".spiceinit"), ".spiceinit"]:
        try:
            shutil.copyfile(spinit_src, target)
        except Exception:
            pass

# 2. Configurar variables de entorno explícitas para Ngspice y Sky130
env = os.environ.copy()
env["HOME"] = "/headless"
env["PDK"] = "sky130A"
env["PDK_ROOT"] = "/foss/pdks"
env["PATH"] = "/foss/tools/bin:" + env.get("PATH", "")

# 3. Exportar netlist actualizado desde Xschem
cmd_xschem = ["/foss/tools/bin/xschem", "-n", "-s", "-r", "-q", "/foss/designs/LDO-sky130/xschem/ldo_tb_montecarlo.sch", "-o", "/headless/.xschem/simulations/"]
try:
    subprocess.run(cmd_xschem, capture_output=True, text=True, env=env)
except Exception as e:
    print(f"Aviso al generar netlist con xschem: {e}")

# 4. Ejecutar Ngspice
print("Ejecutando simulaciones Monte Carlo en Ngspice...")
t0 = time.time()
comando = ["/foss/tools/bin/ngspice", "-b", netlist_path]
if not os.path.exists(comando[0]):
    comando[0] = "ngspice"
resultado = subprocess.run(comando, capture_output=True, text=True, cwd=raw_path, env=env)
t1 = time.time()
mi_log = (resultado.stdout or "") + "\n" + (resultado.stderr or "")

if resultado.returncode != 0 or "could not find a valid modelname" in mi_log.lower():
    print("La simulación falló. Aquí está el error de Ngspice:\n")
    print(resultado.stderr or resultado.stdout)
else:
    print(f"¡Simulación completada con éxito en {t1-t0:.1f} segundos!")
```

```python
import re
import pandas as pd

# 1. Separar el log por cada iteración del bucle Monte Carlo
bloques = re.split(r"CORRIENDO MONTE CARLO ITERACION:\s*\d+", mi_log)
patron = r"([a-zA-Z0-9_]+)\s*=\s*([+-]?\d+\.?\d*[eE]?[+-]?\d*)"

lista_iteraciones = []
for i, bloque in enumerate(bloques[1:], start=1):
    coincidencias = re.findall(patron, bloque)
    d = {"iteracion": i}
    for nombre, valor in coincidencias:
        d[nombre.strip().lower()] = float(valor)
    d.pop("tnom", None)
    lista_iteraciones.append(d)

# 2. Convertir la lista de las 10 iteraciones a un DataFrame de pandas
df_montecarlo = pd.DataFrame(lista_iteraciones)
print(f"Se extrajeron con éxito {len(df_montecarlo)} iteraciones Monte Carlo:")
display(df_montecarlo)
```

```python
# =========================================================
# CELDA 4: HISTOGRAMAS Y ANÁLISIS ESTADÍSTICO DE MONTE CARLO
# =========================================================
import matplotlib.pyplot as plt
import numpy as np

# Configuración de figura de 3x3 subplots
fig, axes = plt.subplots(3, 3, figsize=(15, 11))
fig.suptitle("Distribución de Histogramas - Monte Carlo (LDO Sky130)", fontsize=15, fontweight="bold")

# Definición de métricas requeridas
metricas = [
    ("undershoot_mv", "Undershoot (mV)", "coral"),
    ("overshoot_mv", "Overshoot (mV)", "mediumpurple"),
    ("psrr_1", "PSRR @ 1Hz (dB)", "teal"),
    ("margen_fase", "Margen de Fase (°)", "limegreen"),
    ("line_regulation", "Line Regulation (mV)", "dodgerblue"),
    ("load_regulation", "Load Regulation (mV)", "indianred"),
    ("dropout_mv", "Dropout Voltage (mV)", "goldenrod"),
    ("ugb", "UGB (Hz)", "darkcyan"),
    ("ganancia_banda_plana", "Ganancia Banda Plana (dB)", "mediumvioletred")
] 

for idx, (col, titulo, color) in enumerate(metricas):
    row = idx // 3
    c = idx % 3
    ax = axes[row, c]
    if col in df_montecarlo.columns:
        data = df_montecarlo[col].dropna()
        ax.hist(data, bins=6, color=color, edgecolor="black", alpha=0.7)
        mean_val = data.mean()
        std_val = data.std()
        ax.axvline(mean_val, color="red", linestyle="--", linewidth=1.5, label=f"Media: {mean_val:.2f}")
        ax.set_title(f"{titulo}\nMean: {mean_val:.2f} | Std: {std_val:.2f}", fontsize=10, fontweight="bold")
        ax.set_ylabel("Frecuencia")
        ax.legend(fontsize=8)
        ax.grid(True, linestyle=":", alpha=0.6)

plt.tight_layout()
plt.show()
```
