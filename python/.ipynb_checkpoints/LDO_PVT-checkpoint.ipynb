```python
import re
import os
import subprocess
from PyLTSpice import RawRead

parametros_ldo = {}
graficas_ldo = {}

def one_pvt1hilo(netlist_path, temp, vdd, proces):
    print(f"Ejecutando corner: temperatura = {temp}, VDD = {vdd}, Proceso = {proces}")
    
    # 1. Extraemos la carpeta original donde vive el .spice
    directorio_sim = os.path.dirname(netlist_path)
    
    # 2. Leer y modificar el netlist
    with open(netlist_path, 'r') as archivo:
        contenido = archivo.read()

    patron_corner = r"(\.lib\s+/foss/pdks/sky130A/libs\.tech/ngspice/sky130\.lib\.spice\s+)\w+"
    contenido = re.sub(patron_corner, rf"\g<1>{proces}", contenido)
    
    patron_temp = r"(\.option\s+temp=)[-\d\.]+"
    contenido = re.sub(patron_temp, rf"\g<1>{temp}", contenido)
    
    patron_vdd = r"(\.param\s+VDD\s*=\s*)[-\d\.]+"
    contenido = re.sub(patron_vdd, rf"\g<1>{vdd}", contenido)
    
    with open(netlist_path, 'w') as archivo:
        archivo.write(contenido)

    # 3. Simular forzando a Ngspice a quedarse en su carpeta natural
    comando = ['/foss/tools/ngspice/bin/ngspice', "-b", netlist_path]
    resultado = subprocess.run(comando, capture_output=True, text=True, cwd=directorio_sim)
    
    if resultado.returncode != 0:
        print("La simulación falló. Aquí está el error de Ngspice:\n")
        print(resultado.stderr)
        return
    
    # 4. Extraer el LOG
    mi_log = resultado.stdout
    patron = r"([a-zA-Z0-9_]+)\s*=\s*([+-]?\d+\.?\d*[eE]?[+-]?\d*)"
    coincidencias = re.findall(patron, mi_log)
    
    nombre_llave = f"{vdd}[V]_{temp}[C]_{proces}"
    
    parametros_ldo[nombre_llave] = {}
    graficas_ldo[nombre_llave] = {}

    for nombre, valor in coincidencias:
        parametros_ldo[nombre_llave][nombre.strip().lower()] = float(valor)
    
    if "tnom" in parametros_ldo[nombre_llave]:
        del parametros_ldo[nombre_llave]["tnom"]

    # 5. Extraer vectores apuntando al mismo directorio_sim (donde cayeron los .raw)
    raw_tran = RawRead(os.path.join(directorio_sim, "LDO_tran_picos.raw"))
    raw_ac_stability = RawRead(os.path.join(directorio_sim, "LDO_ac_estability.raw"))
    raw_ac_psrr = RawRead(os.path.join(directorio_sim, "LDO_ac_psrr.raw"))
    raw_dc_line = RawRead(os.path.join(directorio_sim, "LDO_dc_line.raw"))
    raw_dc_load = RawRead(os.path.join(directorio_sim, "LDO_dc_load.raw"))
    
    graficas_ldo[nombre_llave]["time"] = raw_tran.get_trace("time").get_wave()
    graficas_ldo[nombre_llave]["t_vo"] = raw_tran.get_trace("v(vo)").get_wave()
    
    graficas_ldo[nombre_llave]["frecuencia"] = raw_ac_stability.get_trace("frequency").get_wave()
    graficas_ldo[nombre_llave]["fase"] = raw_ac_stability.get_trace("fase_deg").get_wave()
    graficas_ldo[nombre_llave]["ganancia"] = raw_ac_stability.get_trace("loop_gain_db").get_wave()
    
    graficas_ldo[nombre_llave]["psrr"] = raw_ac_psrr.get_trace("psrr_db").get_wave()
    
    graficas_ldo[nombre_llave]["dc_vin"] = raw_dc_line.get_trace("v(v-sweep)").get_wave()
    graficas_ldo[nombre_llave]["dc_vin_vo"] = raw_dc_line.get_trace("v(vo)").get_wave()
    
    graficas_ldo[nombre_llave]["dc_iload"] = raw_dc_load.get_trace("i(i-sweep)").get_wave()
    graficas_ldo[nombre_llave]["dc_iload_vo"] = raw_dc_load.get_trace("v(vo)").get_wave()
    
    print("¡Extracción completada!")


# ==============================================================================
# BUCLE DE EJECUCIÓN
# ==============================================================================
temperaturas = [-40, 27, 125]
voltajes = [1.62, 1.8, 1.98] 
corners_proceso = [
    "tt", "sf", "ff", "ss", "fs", "ll", "hh", "hl", "lh",
    "sf_ll", "sf_hh", "sf_hl", "sf_lh", "ff_ll", "ff_hh", "ff_hl", "ff_lh",
    "ss_ll", "ss_hh", "ss_hl", "ss_lh", "fs_ll", "fs_hh", "fs_hl", "fs_lh"
]

netlist_path = "/headless/.xschem/simulations/ldo_tb.spice"

total_sim = len(temperaturas) * len(voltajes) * len(corners_proceso)
contador = 1

print(f"Iniciando barrido secuencial PVT: {total_sim} simulaciones programadas.\n")

for temp in temperaturas:
    for vdd in voltajes:
        for proces in corners_proceso:
            print(f"--- Progreso: {contador}/{total_sim} ---")
            
            # Ya no pasamos raw_path, la función se encarga de todo
            one_pvt1hilo(netlist_path, temp, vdd, proces)
            
            contador += 1

print("\n¡Barrido masivo finalizado con éxito!")
```

```python
import re
import os
import subprocess
from PyLTSpice import RawRead

parametros_ldo = {}
graficas_ldo = {}

def one_pvt(netlist_path, raw_path, temp, vdd, proces):
    print(f"Ejecutando corner: temperatura = {temp}, VDD = {vdd}, Proceso = {proces}")
    
    # ====================================================================
    # PASO 1 y 2: LEER Y MODIFICAR EL CONTENIDO 
    # ====================================================================
    with open(netlist_path, 'r') as archivo:
        contenido = archivo.read() # Corregida la indentación

    patron_corner = r"(\.lib\s+/foss/pdks/sky130A/libs\.tech/ngspice/sky130\.lib\.spice\s+)\w+"
    contenido = re.sub(patron_corner, rf"\g<1>{proces}", contenido)
    
    patron_temp = r"(\.option\s+temp=)[-\d\.]+"
    contenido = re.sub(patron_temp, rf"\g<1>{temp}", contenido)
    
    patron_vdd = r"(\.param\s+VDD\s*=\s*)[-\d\.]+"
    contenido = re.sub(patron_vdd, rf"\g<1>{vdd}", contenido)
    
    # ====================================================================
    # PASO 3: SOBRESCRIBIR EL ARCHIVO
    # ====================================================================
    with open(netlist_path, 'w') as archivo:
        archivo.write(contenido) # Corregida la indentación

    # ====================================================================
    # PASO 4: SIMULAR
    # ====================================================================
    comando = ['/foss/tools/ngspice/bin/ngspice', "-b", netlist_path]
    resultado = subprocess.run(comando, capture_output=True, text=True, cwd=raw_path)
    mi_log = resultado.stdout
    
    if resultado.returncode != 0:
        print("La simulación falló. Aquí está el error de Ngspice:\n")
        print(resultado.stderr)
        return # Sale de la función si hay error para no romper el código
    else:
        print("¡Simulación terminada con éxito!")
    
    # ====================================================================
    # PASO 5: EXTRAER PARÁMETROS DEL LOG
    # ====================================================================
    patron = r"([a-zA-Z0-9_]+)\s*=\s*([+-]?\d+\.?\d*[eE]?[+-]?\d*)"
    coincidencias = re.findall(patron, mi_log)
    
    nombre_llave = f"{vdd}[V]_{temp}[C]_{proces}" # Variables corregidas
    
    # Inicializar los diccionarios anidados para esta iteración
    parametros_ldo[nombre_llave] = {}
    graficas_ldo[nombre_llave] = {}

    for nombre, valor in coincidencias:
        parametros_ldo[nombre_llave][nombre.strip().lower()] = float(valor)
    
    # Validar que 'tnom' exista antes de intentar borrarlo
    if "tnom" in parametros_ldo[nombre_llave]:
        del parametros_ldo[nombre_llave]["tnom"]

    # ====================================================================
    # PASO 6: EXTRAER VECTORES DE LOS .RAW
    # ====================================================================
    raw_tran = RawRead(os.path.join(raw_path,"LDO_tran_picos.raw"))
    raw_ac_stability = RawRead(os.path.join(raw_path,"LDO_ac_estability.raw"))
    raw_ac_psrr = RawRead(os.path.join(raw_path,"LDO_ac_psrr.raw"))
    raw_dc_line = RawRead(os.path.join(raw_path,"LDO_dc_line.raw"))
    raw_dc_load = RawRead(os.path.join(raw_path,"LDO_dc_load.raw"))
    
    # Uso de comillas en las llaves y extracción del raw correcto para cada análisis
    graficas_ldo[nombre_llave]["time"] = raw_tran.get_trace("time").get_wave()
    graficas_ldo[nombre_llave]["t_vo"] = raw_tran.get_trace("v(vo)").get_wave()
    
    graficas_ldo[nombre_llave]["frecuencia"] = raw_ac_stability.get_trace("frequency").get_wave()
    graficas_ldo[nombre_llave]["fase"] = raw_ac_stability.get_trace("fase_deg").get_wave()
    graficas_ldo[nombre_llave]["ganancia"] = raw_ac_stability.get_trace("loop_gain_db").get_wave()
    
    graficas_ldo[nombre_llave]["psrr"] = raw_ac_psrr.get_trace("psrr_db").get_wave()
    
    graficas_ldo[nombre_llave]["dc_vin"] = raw_dc_line.get_trace("v(v-sweep)").get_wave()
    graficas_ldo[nombre_llave]["dc_vin_vo"] = raw_dc_line.get_trace("v(vo)").get_wave()
    
    graficas_ldo[nombre_llave]["dc_iload"] = raw_dc_load.get_trace("i(i-sweep)").get_wave()
    graficas_ldo[nombre_llave]["dc_iload_vo"] = raw_dc_load.get_trace("v(vo)").get_wave()
```

```python
# 1. Definir los parámetros para el barrido PVT
temperaturas = [-40, 27, 125]
voltajes = [1.62, 1.8, 1.98] 
corners_proceso = [
    "tt", "sf", "ff", "ss", "fs",
    "ll", "hh", "hl", "lh",
    "sf_ll", "sf_hh", "sf_hl", "sf_lh",
    "ff_ll", "ff_hh", "ff_hl", "ff_lh",
    "ss_ll", "ss_hh", "ss_hl", "ss_lh",
    "fs_ll", "fs_hh", "fs_hl", "fs_lh"
]

# Rutas originales
netlist_path = "/headless/.xschem/simulations/ldo_tb.spice"
raw_path = "/foss/designs/LDO-sky130/xschem/spice_raw_log/raw/"

# 2. Inicializar variables de seguimiento
total_sim = len(temperaturas) * len(voltajes) * len(corners_proceso)
contador = 1

print(f"Iniciando barrido secuencial PVT: {total_sim} simulaciones programadas.\n")

# 3. Bucles anidados para iterar sobre todas las combinaciones posibles (1 hilo)
for temp in temperaturas:
    for vdd in voltajes:
        for proces in corners_proceso:
            print(f"--- Progreso: {contador}/{total_sim} ---")
            
            # Llamar a tu función de un solo hilo
            one_pvt1hilo(netlist_path, raw_path, temp, vdd, proces)
            
            contador += 1

print("\n¡Barrido masivo finalizado con éxito!")
```

```python
import pandas as pd
import plotly.graph_objects as go

# ====================================================================
# 1. MOSTRAR LA TABLA DE RESULTADOS (LOG)
# ====================================================================
# Pandas convierte automáticamente el diccionario anidado en una tabla
df_resultados = pd.DataFrame.from_dict(parametros_ldo, orient='index')

print("--- Tabla de Parámetros PVT ---")
# Si usas Jupyter Notebook, 'display' dibuja una tabla interactiva muy limpia. 
# Si usas consola plana, cambia 'display' por 'print'
display(df_resultados) 

# ====================================================================
# 2. GRAFICAR TODAS LAS CURVAS EN UNA SOLA FIGURA (Ej: PSRR)
# ====================================================================
fig_psrr = go.Figure()

# Iteramos sobre el diccionario principal. 
# 'llave' será el nombre (ej. 1.8[V]_27[C]_tt)
# 'curvas' contendrá los vectores asociados a esa llave.
for llave, curvas in graficas_ldo.items():
    fig_psrr.add_trace(go.Scatter(
        x=curvas["frecuencia"],  # Eje X
        y=curvas["psrr"],        # Eje Y
        mode='lines',
        name=llave               # Este nombre aparecerá en la leyenda
    ))

# Configuramos el diseño de la gráfica (título, etiquetas y escala)
fig_psrr.update_layout(
    title="Análisis PVT: PSRR del LDO",
    xaxis_title="Frecuencia [Hz]",
    yaxis_title="PSRR [dB]",
    xaxis_type="log",            # Escala logarítmica para frecuencia
    legend_title="Corners (VDD_Temp_Proceso)",
    template="plotly_dark"       # Tema oscuro (opcional, acorde a Xschem)
)

fig_psrr.show()

# ====================================================================
# 3. GRAFICAR OTRA VARIABLE (Ej: Respuesta Transitoria)
# ====================================================================
fig_tran = go.Figure()

for llave, curvas in graficas_ldo.items():
    fig_tran.add_trace(go.Scatter(
        x=curvas["time"], 
        y=curvas["t_vo"], 
        mode='lines',
        name=llave
    ))

fig_tran.update_layout(
    title="Análisis PVT: Respuesta Transitoria (Undershoot/Overshoot)",
    xaxis_title="Tiempo [s]",
    yaxis_title="Voltaje de Salida (Vo) [V]",
    template="plotly_dark"
)

fig_tran.show()
```
