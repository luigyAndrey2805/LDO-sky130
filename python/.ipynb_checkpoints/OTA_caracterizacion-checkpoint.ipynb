# Caracterización Completa y Rango de Salida de la OTA (Sky130)
---
Este notebook realiza la caracterización del Amplificador Operacional de Transconductancia (**OTA**) definido en `OTA_tb.sch`:

1. **Análisis en Frecuencia (Respuesta AC):**
   - Extracción de **Ganancia DC ($A_{v0}$)**, **Ancho de Banda a -3dB ($f_{-3dB}$)**, **Frecuencia de Ganancia Unitaria ($GBW$)** y **Margen de Fase ($PM$)**.
   - Diagrama de Bode Interactivo (Magnitud y Fase) con `plotly.graph_objects`.
2. **Tabla de Punto de Operación (.OP):**
   - Extracción de parámetros $I_D, g_m, g_{ds}, r_o, g_m r_o, V_{dsat}$ para todos los transistores PMOS y NMOS de la OTA.
3. **Simulación de Rango de Salida en Transitorio (Transient Output Swing):**
   - Aplicación de una rampa en el modo común de entrada ($V_{in}$ de $0\text{V}$ a $1.8\text{V}$).
   - Gráfica dinámica de la señal en el tiempo $V_o(t)$ y curva de transferencia $V_o$ vs $V_{in}$ para identificar la excursión máxima de salida ($V_{o,min}$ a $V_{o,max}$).

```python
# ==============================================================================
# Encabezado: Simulación AC y Análisis de Bode de la OTA
# Descripción: Genera el netlist desde Xschem, fuerza formato ASCII en NGspice,
#              ejecuta la simulación AC, carga los datos con spyci, calcula Ganancia DC,
#              GBW y Margen de Fase, y dibuja el Diagrama de Bode con plotly.graph_objects.
# ==============================================================================

import os
import subprocess
import numpy as np
np.complex_ = complex  # Parche de compatibilidad para spyci con NumPy 2.0
from spyci import spyci
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ── 1. Generación del Netlist y Asegurar Formato ASCII ──────────────────────────
print("Generando netlist de OTA_tb.sch...")
cmd_netlist = 'export PATH=/foss/tools/xschem/bin:/foss/tools/ngspice/bin:$PATH; cd ../xschem && xschem -n -q --rcfile /foss/pdks/sky130A/libs.tech/xschem/xschemrc OTA_tb.sch'
subprocess.run(cmd_netlist, shell=True, check=True)

SPICE_AC_FILE = '/headless/.xschem/simulations/OTA_tb.spice'
RAW_AC_FILE   = '/headless/.xschem/simulations/OTA_tb_ac.raw'

with open(SPICE_AC_FILE, 'r') as f:
    orig_net_ac = f.read()

# Configurar set filetype=ascii y ruta RAW específica
net_ac = orig_net_ac.replace(
    'save all',
    'save all\n  set filetype=ascii'
).replace(
    'write OTA_tb.raw all',
    'write OTA_tb_ac.raw all'
)

with open(SPICE_AC_FILE, 'w') as f:
    f.write(net_ac)

print("Ejecutando simulación AC en NGspice...")
cmd_sim_ac = f'export PATH=/foss/tools/xschem/bin:/foss/tools/ngspice/bin:$PATH; cd /headless/.xschem/simulations && ngspice -b OTA_tb.spice'
result_ac = subprocess.run(cmd_sim_ac, shell=True, capture_output=True, text=True)

if result_ac.returncode == 0:
    print("¡Simulación AC completada con éxito!")
else:
    print("Error durante la simulación AC:", result_ac.stderr)

# ── 2. Carga de Datos AC con spyci ─────────────────────────────────────────────
data_ac_raw = spyci.load_raw(RAW_AC_FILE)['values']

freq = np.real(data_ac_raw['frequency'])
vo_ac = data_ac_raw['v(vo)']

# Cálculo de Magnitud (dB) y Fase (grados)
mag_db = 20 * np.log10(np.abs(vo_ac))
phase_deg = np.angle(vo_ac, deg=True)

# Desenvolver fase para evitar saltos
phase_deg = np.unwrap(np.radians(phase_deg)) * (180 / np.pi)

# ── 3. Extracción de Parámetros AC (Ganancia DC, GBW, Margen de Fase) ──────────
dc_gain_db = float(mag_db[0])
dc_gain_linear = 10 ** (dc_gain_db / 20)

# Frecuencia a -3dB
target_3db = dc_gain_db - 3.0
idx_3db = int(np.argmin(np.abs(mag_db - target_3db)))
f_3db = float(freq[idx_3db])

# Frecuencia de Ganancia Unitaria (GBW @ 0 dB)
idx_gbw = int(np.argmin(np.abs(mag_db - 0.0)))
gbw_freq = float(freq[idx_gbw])

# Margen de Fase (PM = 180 + Fase a la frecuencia GBW)
phase_at_gbw = float(phase_deg[idx_gbw])
phase_margin = 180.0 + phase_at_gbw

print(f"=== RESULTADOS DE CARACTERIZACIÓN AC ===")
print(f"Ganancia en DC (Av0):           {dc_gain_db:.2f} dB ({dc_gain_linear:.2f} V/V)")
print(f"Frecuencia de Corte (-3dB):    {f_3db/1e3:.2f} kHz")
print(f"Ancho de Banda GBW (0 dB):     {gbw_freq/1e6:.2f} MHz")
print(f"Margen de Fase (PM):           {phase_margin:.2f}°")

# ── 4. Diagrama de Bode con Plotly ─────────────────────────────────────────────
fig_bode = make_subplots(
    rows=2, cols=1,
    shared_xaxes=True,
    vertical_spacing=0.1,
    subplot_titles=('Respuesta en Magnitud [dB]', 'Respuesta en Fase [grados]')
)

# Traza de Magnitud
fig_bode.add_trace(
    go.Scatter(x=freq, y=mag_db, mode='lines', name='Magnitud [dB]',
               line=dict(color='#1f77b4', width=2.5)),
    row=1, col=1
)

# Traza de Fase
fig_bode.add_trace(
    go.Scatter(x=freq, y=phase_deg, mode='lines', name='Fase [°]',
               line=dict(color='#d62728', width=2.5)),
    row=2, col=1
)

# Marcador GBW
fig_bode.add_trace(
    go.Scatter(x=[gbw_freq], y=[0], mode='markers+text', name='GBW',
               text=[f"GBW: {gbw_freq/1e6:.2f} MHz"], textposition="top right",
               marker=dict(size=10, color='green', symbol='star')),
    row=1, col=1
)

fig_bode.update_xaxes(type="log", title_text="Frecuencia [Hz]", row=2, col=1)
fig_bode.update_yaxes(title_text="Magnitud [dB]", row=1, col=1)
fig_bode.update_yaxes(title_text="Fase [°]", row=2, col=1)

fig_bode.update_layout(
    title_text=f"Diagrama de Bode de la OTA (Av0 = {dc_gain_db:.1f} dB, GBW = {gbw_freq/1e6:.2f} MHz, PM = {phase_margin:.1f}°)",
    height=700, width=1000,
    template="plotly_white"
)

fig_bode.show()
```

```python
# ==============================================================================
# Encabezado: Extracción de Tabla de Punto de Operación (.OP) de Transistores de la OTA
# Descripción: Extrae los parámetros de punto de operación (Id, gm, gds, ro, gm*ro, Vdsat)
#              para todos los transistores PMOS y NMOS pertenecientes a la OTA (subcircuito x1).
# ==============================================================================

import pandas as pd

# Mapeo de transistores de la OTA en el subcircuito x1
ota_transistores = {
    'Mnr (PMOS Espejo Izq)':  {'gm': '@m.x1.xmnr.msky130_fd_pr__pfet_01v8[gm]', 'gds': '@m.x1.xmnr.msky130_fd_pr__pfet_01v8[gds]', 'id': 'i(@m.x1.xmnr.msky130_fd_pr__pfet_01v8[id])', 'vdsat': 'v(@m.x1.xmnr.msky130_fd_pr__pfet_01v8[vdsat])'},
    'Mpr (PMOS Espejo Der)':  {'gm': '@m.x1.xmpr.msky130_fd_pr__pfet_01v8[gm]', 'gds': '@m.x1.xmpr.msky130_fd_pr__pfet_01v8[gds]', 'id': 'i(@m.x1.xmpr.msky130_fd_pr__pfet_01v8[id])', 'vdsat': 'v(@m.x1.xmpr.msky130_fd_pr__pfet_01v8[vdsat])'},
    'Mo  (PMOS Salida)':      {'gm': '@m.x1.xmo.msky130_fd_pr__pfet_01v8[gm]',  'gds': '@m.x1.xmo.msky130_fd_pr__pfet_01v8[gds]',  'id': 'i(@m.x1.xmo.msky130_fd_pr__pfet_01v8[id])',  'vdsat': 'v(@m.x1.xmo.msky130_fd_pr__pfet_01v8[vdsat])'},
    'Mn  (NMOS Par Diff Izq)':{'gm': '@m.x1.xmn.msky130_fd_pr__nfet_01v8[gm]',  'gds': '@m.x1.xmn.msky130_fd_pr__nfet_01v8[gds]',  'id': 'i(@m.x1.xmn.msky130_fd_pr__nfet_01v8[id])',  'vdsat': 'v(@m.x1.xmn.msky130_fd_pr__nfet_01v8[vdsat])'},
    'Mp  (NMOS Par Diff Der)':{'gm': '@m.x1.xmp.msky130_fd_pr__nfet_01v8[gm]',  'gds': '@m.x1.xmp.msky130_fd_pr__nfet_01v8[gds]',  'id': 'i(@m.x1.xmp.msky130_fd_pr__nfet_01v8[id])',  'vdsat': 'v(@m.x1.xmp.msky130_fd_pr__nfet_01v8[vdsat])'},
    'Mr  (NMOS Ref Iref)':    {'gm': '@m.x1.xmr.msky130_fd_pr__nfet_01v8[gm]',  'gds': '@m.x1.xmr.msky130_fd_pr__nfet_01v8[gds]',  'id': 'i(@m.x1.xmr.msky130_fd_pr__nfet_01v8[id])',  'vdsat': 'v(@m.x1.xmr.msky130_fd_pr__nfet_01v8[vdsat])'},
    'Mrc1 (NMOS Tail/Cola)':  {'gm': '@m.x1.xmrc1.msky130_fd_pr__nfet_01v8[gm]','gds': '@m.x1.xmrc1.msky130_fd_pr__nfet_01v8[gds]','id': 'i(@m.x1.xmrc1.msky130_fd_pr__nfet_01v8[id])','vdsat': 'v(@m.x1.xmrc1.msky130_fd_pr__nfet_01v8[vdsat])'},
    'Mc2 (NMOS Salida Inf)':  {'gm': '@m.x1.xmc2.msky130_fd_pr__nfet_01v8[gm]', 'gds': '@m.x1.xmc2.msky130_fd_pr__nfet_01v8[gds]', 'id': 'i(@m.x1.xmc2.msky130_fd_pr__nfet_01v8[id])', 'vdsat': 'v(@m.x1.xmc2.msky130_fd_pr__nfet_01v8[vdsat])'}
}

# Índice 0 para el punto de operación DC OP
idx_op = 0
tabla_ota_op = []

for nombre, keys in ota_transistores.items():
    gm = float(np.real(abs(data_ac_raw[keys['gm']][idx_op])))
    gds = float(np.real(abs(data_ac_raw[keys['gds']][idx_op])))
    id_val = float(np.real(abs(data_ac_raw[keys['id']][idx_op])))
    vdsat = float(np.real(data_ac_raw[keys['vdsat']][idx_op]))
    
    ro = 1.0 / gds if gds != 0 else np.nan
    gm_id = gm / id_val if id_val != 0 else np.nan
    gm_ro = gm * ro

    tabla_ota_op.append({
        'Transistor': nombre,
        'Id (uA)': round(id_val * 1e6, 2),
        'gm (uS)': round(gm * 1e6, 2),
        'gds (uS)': round(gds * 1e6, 2),
        'ro (kOhm)': round(ro / 1e3, 2),
        'gm*ro (V/V)': round(gm_ro, 2),
        'gm/Id (1/V)': round(gm_id, 2),
        'Vdsat (V)': round(vdsat, 3)
    })

df_ota_op = pd.DataFrame(tabla_ota_op)
print("=== TABLA DE PUNTO DE OPERACIÓN (.OP) DE LA OTA ===")
df_ota_op
```

```python
# ==============================================================================
# Encabezado: Simulación de Rango de Salida en Transitorio (Output Voltage Swing)
# Descripción: Modifica el netlist para aplicar una rampa en la tensión de entrada
#              de modo común (V2) desde un valor por debajo del modo común (0V)
#              hasta por encima (1.8V). Ejecuta un análisis transitorio (tran 10u 10m)
#              y grafica Vo(t) y la característica de transferencia Vo vs Vin con Plotly.
# ==============================================================================

import plotly.graph_objects as go
from plotly.subplots import make_subplots

# ── 1. Configuración de la Rampa Transitoria ──────────────────────────────────
VCM_START = 0.0   # Tensión inicial de rampa [V]
VCM_STOP  = 1.8   # Tensión final de rampa [V]
T_SIM     = '10m' # Tiempo total de simulación transitoria
T_STEP    = '10u' # Paso de tiempo

RAW_TRAN_FILE = '/headless/.xschem/simulations/OTA_tb_tran.raw'
SPICE_TRAN_FILE = '/foss/designs/LDO-sky130/xschem/OTA_tb_tran.spice'

with open(SPICE_AC_FILE, 'r') as f:
    orig_net = f.read()

# Modificar el netlist para análisis transitorio con rampa PWL y formato ASCII
tran_net = orig_net.replace(
    'V2 net3 0 {VCM}',
    f'V2 net3 0 PWL(0 {VCM_START} {T_SIM} {VCM_STOP})'
).replace(
    'ac dec 10 1 1G',
    f'tran {T_STEP} {T_SIM}'
).replace(
    'plot vdb(vo)',
    '* plot vdb(vo)'
).replace(
    'plot cph(v(vo))*180/PI',
    '* plot cph(v(vo))*180/PI'
).replace(
    'OTA_tb_ac.raw',
    'OTA_tb_tran.raw'
)

with open(SPICE_TRAN_FILE, 'w') as f:
    f.write(tran_net)

print(f"Lanzando simulación transitoria de rampa de salida (PWL {VCM_START}V a {VCM_STOP}V)...")
cmd_sim_tran = f'export PATH=/foss/tools/xschem/bin:/foss/tools/ngspice/bin:$PATH; cd /headless/.xschem/simulations && ngspice -b /foss/designs/LDO-sky130/xschem/OTA_tb_tran.spice'
result_tran = subprocess.run(cmd_sim_tran, shell=True, capture_output=True, text=True)

if result_tran.returncode == 0:
    print("¡Simulación transitoria completada con éxito!")
else:
    print("Error durante la simulación transitoria:", result_tran.stderr)

# ── 2. Carga y Extracción de Datos Transitorios ───────────────────────────────
data_tran_raw = spyci.load_raw(RAW_TRAN_FILE)['values']

time_arr = np.real(data_tran_raw['time']) * 1e3  # Convertir a ms
vo_tran  = np.real(data_tran_raw['v(vo)'])
vin_tran = np.real(data_tran_raw['v(vip)'])

# Identificación de los límites del Rango de Salida (Output Swing)
vo_min = float(np.min(vo_tran))
vo_max = float(np.max(vo_tran))
output_swing = vo_max - vo_min

print(f"=== RESULTADOS DE RANGO DE SALIDA EN TRANSITORIO ===")
print(f"Tensión Mínima de Salida (Vo,min):  {vo_min:.3f} V")
print(f"Tensión Máxima de Salida (Vo,max):  {vo_max:.3f} V")
print(f"Excursión Máxima de Salida (ΔVo):    {output_swing:.3f} V")

# ── 3. Gráficas de Rango de Salida en Transitorio con Plotly ──────────────────
fig_tran = make_subplots(
    rows=1, cols=2,
    subplot_titles=('Respuesta Transitoria: Vo(t) y Vin(t) vs Tiempo', 'Característica de Transferencia: Vo vs Vin')
)

# Gráfica 1: Vo(t) y Vin(t) vs Tiempo
fig_tran.add_trace(
    go.Scatter(x=time_arr, y=vin_tran, mode='lines', name='Vin (Rampa)',
               line=dict(color='#ff7f0e', width=2, dash='dash')),
    row=1, col=1
)
fig_tran.add_trace(
    go.Scatter(x=time_arr, y=vo_tran, mode='lines', name='Vo (Salida OTA)',
               line=dict(color='#1f77b4', width=2.5)),
    row=1, col=1
)

# Gráfica 2: Vo vs Vin (Transfer Curve)
fig_tran.add_trace(
    go.Scatter(x=vin_tran, y=vo_tran, mode='lines', name='Vo vs Vin',
               line=dict(color='#2ca02c', width=2.5)),
    row=1, col=2
)

# Líneas de referencia para límites de salida en la gráfica 2
fig_tran.add_hline(y=vo_max, line_dash="dot", line_color="red", annotation_text=f"Vo,max = {vo_max:.2f}V", row=1, col=2)
fig_tran.add_hline(y=vo_min, line_dash="dot", line_color="red", annotation_text=f"Vo,min = {vo_min:.2f}V", row=1, col=2)

fig_tran.update_xaxes(title_text="Tiempo [ms]", row=1, col=1)
fig_tran.update_yaxes(title_text="Tensión [V]", row=1, col=1)

fig_tran.update_xaxes(title_text="Tensión de Entrada Vin [V]", row=1, col=2)
fig_tran.update_yaxes(title_text="Tensión de Salida Vo [V]", row=1, col=2)

fig_tran.update_layout(
    title_text=f"Caracterización del Rango de Salida de la OTA (Vo,min = {vo_min:.2f}V, Vo,max = {vo_max:.2f}V, Swing = {output_swing:.2f}V)",
    height=550, width=1200,
    template="plotly_white"
)

fig_tran.show()
```
