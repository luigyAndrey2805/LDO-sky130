# Caracterización y Barrido de Parámetros del LDO Sky130
Este cuaderno permite simular el circuito LDO (Low-DropOut regulator) variando dimensiones físicas ($L, W$) de resistencias, capacitores y componentes de carga.
Las simulaciones se ejecutan en **paralelo con múltiples hilos** usando NGspice en Docker para maximizar la velocidad.

### Funcionalidades:
1. **Barrido de Componentes**:
   - $L, W$ de la Resistencia de Compensación de la OTA (`XR1_OTA`).
   - $W, L, MF$ (multiplicador) del Capacitor MIM de Compensación de la OTA (`XC1_OTA`).
   - Capacitancia de Carga del LDO ($C_L$) y Resistencia de Carga ($R_L$).
2. **Ejecución Paralela Multihilo** con `ThreadPoolExecutor`.
3. **Métricas automáticas**: Ganancia DC (dB), Ancho de Banda $f_{-3dB}$ (Hz), UGB (Hz) y Margen de Fase (°).
4. **Visualización de Resultados**: Familias de curvas de Bode y gráficos de tendencia.



```python
# ==============================================================================
# ENCABEZADO: Configuración e Importaciones
# DESCRIPCIÓN: Importación de librerías para manejo de datos, gráficos y subprocesos.
# ==============================================================================

import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import concurrent.futures
import subprocess
import shutil

# Configuración de estilo visual para los gráficos
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['font.sans-serif'] = 'Inter'
plt.rcParams['axes.edgecolor'] = '#cccccc'
plt.rcParams['axes.linewidth'] = 0.8

print("✓ Librerías cargadas correctamente.")

```

```python
# ==============================================================================
# ENCABEZADO: Lector de archivos RAW de NGspice
# DESCRIPCIÓN: Extrae vectores de frecuencia, magnitud y fase de archivos .raw
# ==============================================================================

def leer_raw_ac(filepath):
    with open(filepath, 'r') as f:
        content = f.read()
    
    vo_idx = None
    lines = content.split('\n')
    for line in lines:
        if 'v(vo)' in line:
            vo_idx = int(line.strip().split()[0])
            break
            
    if vo_idx is None:
        vo_idx = 37
        
    values_sec = content.split("Values:\n")[1]
    raw_lines = values_sec.strip().split("\n")
    
    freq_list = []
    vo_complex_list = []
    
    curr_freq = None
    curr_var_idx = 0
    
    for l in raw_lines:
        parts = l.strip().split()
        if not parts:
            continue
        if len(parts) >= 2 and parts[0].isdigit():
            curr_var_idx = int(parts[0])
            if curr_var_idx == 0:
                curr_freq = float(parts[1])
        elif len(parts) == 1 or "," in parts[0] or (len(parts) == 2 and not parts[0].isdigit()):
            comp_str = "".join(parts)
            c_parts = comp_str.split(",")
            if len(c_parts) == 2:
                try:
                    r_val = float(c_parts[0])
                    i_val = float(c_parts[1])
                    if curr_var_idx == 0:
                        freq_list.append(curr_freq)
                    elif curr_var_idx == vo_idx:
                        vo_complex_list.append(complex(r_val, i_val))
                except ValueError:
                    pass
                    
    freqs = np.array(freq_list[:len(vo_complex_list)])
    vo_c = np.array(vo_complex_list)
    
    mag_db = 20 * np.log10(np.abs(vo_c) + 1e-18)
    phase_deg = np.angle(vo_c, deg=True)
    
    return freqs, mag_db, phase_deg

```

```python
# ==============================================================================
# ENCABEZADO: Generador de Netlist con Parámetros Modificados
# DESCRIPCIÓN: Modifica la subcircuito o parámetros de SPICE para hacer barridos.
# ==============================================================================

BASE_SPICE_PATH = "/home/luigy/eda/designs/LDO-sky130/xschem/ldo_tb.spice"

def crear_netlist_simulacion(target_component, param_name, valor, out_filename):
    with open(BASE_SPICE_PATH, 'r') as f:
        content = f.read()
        
    if target_component == 'XR1_OTA':
        if param_name == 'L':
            content = re.sub(r'(XR1 VO_ota1 net1 VDD sky130_fd_pr__res_high_po W=[\d\.]+ L=)[\d\.]+', f'\\g<1>{valor}', content)
        elif param_name == 'W':
            content = re.sub(r'(XR1 VO_ota1 net1 VDD sky130_fd_pr__res_high_po W=)[\d\.]+', f'\\g<1>{valor}', content)
            
    elif target_component == 'XC1_OTA':
        if param_name == 'W':
            content = re.sub(r'(XC1 VO_ota net1 sky130_fd_pr__cap_mim_m3_2 W=)[\d\.]+', f'\\g<1>{valor}', content)
        elif param_name == 'L':
            content = re.sub(r'(XC1 VO_ota net1 sky130_fd_pr__cap_mim_m3_2 W=[\d\.]+ L=)[\d\.]+', f'\\g<1>{valor}', content)
        elif param_name == 'MF':
            content = re.sub(r'(XC1 VO_ota net1 sky130_fd_pr__cap_mim_m3_2 W=[\d\.]+ L=[\d\.]+ MF=)[\d\.]+', f'\\g<1>{valor}', content)
            
    elif target_component == 'CL':
        content = re.sub(r'(\.param CL =) [\d\.\w]+', f'\\g<1>{valor}', content)
    elif target_component == 'RL':
        content = re.sub(r'(\.param RL =) [\d\.\w]+', f'\\g<1>{valor}', content)
        
    with open(out_filename, 'w') as f:
        f.write(content)

```

```python
# ==============================================================================
# ENCABEZADO: Motor de Simulación Paralela Multihilo
# DESCRIPCIÓN: Corre múltiples instancias de NGspice concurrentemente en Docker.
# ==============================================================================

def simular_instancia(args):
    idx, target_comp, param_name, val = args
    
    host_tmp_dir = f"/home/luigy/eda/designs/LDO-sky130/python/sim_tmp/run_{idx}"
    docker_tmp_dir = f"/foss/designs/LDO-sky130/python/sim_tmp/run_{idx}"
    os.makedirs(host_tmp_dir, exist_ok=True)
    
    spice_path = os.path.join(host_tmp_dir, "ldo_tb.spice")
    raw_path = os.path.join(host_tmp_dir, "LDO_ac.raw")
    
    crear_netlist_simulacion(target_comp, param_name, val, spice_path)
    
    docker_cmd = f'docker exec iic-osic-tools_chipathon_xvnc_uid_1000 bash -c "cd {docker_tmp_dir} && /foss/tools/bin/ngspice -b ldo_tb.spice"'
    
    proc = subprocess.run(docker_cmd, shell=True, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)
    output = proc.stdout
    
    f3db = None
    ugb = None
    fase_ugb = None
    margen_fase = None
    
    for line in output.split("\n"):
        if "f3db =" in line:
            try: f3db = float(line.split("=")[1].strip())
            except: pass
        elif "ugb =" in line:
            try: ugb = float(line.split("=")[1].strip())
            except: pass
        elif "fase_ugb =" in line:
            try: fase_ugb = float(line.split("=")[1].strip())
            except: pass
        elif "margen_fase =" in line:
            try: margen_fase = float(line.split("=")[1].strip())
            except: pass

    freqs, mag_db, phase_deg = None, None, None
    if os.path.exists(raw_path):
        try:
            freqs, mag_db, phase_deg = leer_raw_ac(raw_path)
            if ugb is None or margen_fase is None:
                zero_cross_idx = np.where(np.diff(np.sign(mag_db)))[0]
                if len(zero_cross_idx) > 0:
                    i_c = zero_cross_idx[0]
                    ugb = float(freqs[i_c])
                    fase_at_ugb = float(phase_deg[i_c])
                    margen_fase = float(180.0 + fase_at_ugb)
                    f3db = float(freqs[np.argmin(np.abs(mag_db - (mag_db[0] - 3.0)))])
        except Exception:
            pass

    return {
        'idx': idx,
        'val': val,
        'f3db': f3db,
        'ugb': ugb,
        'margen_fase': margen_fase,
        'freqs': freqs,
        'mag_db': mag_db,
        'phase_deg': phase_deg
    }

def ejecutar_barrido(target_comp, param_name, valores, max_workers=6):
    print(f"🚀 Iniciando barrido paralelo para {target_comp} ({param_name}) con {len(valores)} puntos en {max_workers} hilos...\n")
    
    tasks = [(i, target_comp, param_name, val) for i, val in enumerate(valores)]
    results = []
    
    with concurrent.futures.ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = [executor.submit(simular_instancia, task) for task in tasks]
        for future in concurrent.futures.as_completed(futures):
            res = future.result()
            results.append(res)
            if res['margen_fase'] is not None:
                print(f"  [✓] {target_comp}.{param_name} = {res['val']:<8} | MF: {res['margen_fase']:.2f}° | UGB: {res['ugb']:.2e} Hz")
            else:
                print(f"  [!] {target_comp}.{param_name} = {res['val']:<8} | Sin convergencia")
            
    results.sort(key=lambda x: x['idx'])
    print("\n✓ Barrido completado con éxito.")
    return results

```

```python
# ==============================================================================
# ENCABEZADO: Configuración de Parámetros a Barrer
# DESCRIPCIÓN: Selecciona el componente y la lista de valores a simular.
# ==============================================================================

# Opciones de TARGET_COMPONENT:
# - 'XR1_OTA': Resistencia de compensación en la OTA (PARAM_NAME = 'L' o 'W')
# - 'XC1_OTA': Capacitor MIM de compensación en la OTA (PARAM_NAME = 'W', 'L' o 'MF')
# - 'CL': Capacitancia de Carga del LDO (PARAM_NAME = 'value', ej. '0.1u', '1u', '10u')
# - 'RL': Resistencia de Carga del LDO (PARAM_NAME = 'value', ej. 10, 50, 100, 500)

TARGET_COMPONENT = 'XR1_OTA'  # Componente objetivo a evaluar
PARAM_NAME = 'L'             # Parámetro/Dimensión ('L', 'W', 'MF')

# Lista de valores a barrer (puedes modificar este arreglo según lo requieras)
VALORES_A_BARRER = [10.0, 15.0, 22.8, 30.0, 40.0, 50.0]

# Número de hilos de ejecución simultánea (ajustable según núcleos del CPU)
NUM_HILOS_PARALELOS = 6

# Ejecutar la simulación multihilo
resultados = ejecutar_barrido(TARGET_COMPONENT, PARAM_NAME, VALORES_A_BARRER, max_workers=NUM_HILOS_PARALELOS)

```

```python
# ==============================================================================
# ENCABEZADO: Generación de Gráficos y Tabla de Resultados
# DESCRIPCIÓN: Visualiza tendencias de Margen de Fase, UGB y familias Bode.
# ==============================================================================

tabla_data = []
for r in resultados:
    tabla_data.append({
        f'{TARGET_COMPONENT}_{PARAM_NAME}': r['val'],
        'Ganancia DC (dB)': round(r['mag_db'][0], 2) if r['mag_db'] is not None else None,
        'f_-3dB (Hz)': f"{r['f3db']:.2e}" if r['f3db'] else None,
        'UGB (Hz)': f"{r['ugb']:.2e}" if r['ugb'] else None,
        'Margen de Fase (°)': round(r['margen_fase'], 2) if r['margen_fase'] else None
    })

df_res = pd.DataFrame(tabla_data)
display(df_res)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

x_vals = [r['val'] for r in resultados]
pm_vals = [r['margen_fase'] for r in resultados]
ugb_vals = [r['ugb'] for r in resultados]

ax1.plot(x_vals, pm_vals, 'o-', color='#1f77b4', linewidth=2, markersize=8)
ax1.set_title(f'Margen de Fase vs {TARGET_COMPONENT} ({PARAM_NAME})', fontsize=12, fontweight='bold')
ax1.set_xlabel(f'{PARAM_NAME} de {TARGET_COMPONENT}')
ax1.set_ylabel('Margen de Fase (°)')
ax1.axhline(60, color='green', linestyle='--', alpha=0.7, label='MF Óptimo = 60°')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(x_vals, ugb_vals, 's-', color='#ff7f0e', linewidth=2, markersize=8)
ax2.set_title(f'UGB (Ancho de Banda) vs {TARGET_COMPONENT} ({PARAM_NAME})', fontsize=12, fontweight='bold')
ax2.set_xlabel(f'{PARAM_NAME} de {TARGET_COMPONENT}')
ax2.set_ylabel('UGB (Hz)')
ax2.set_yscale('log')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

fig, (ax_mag, ax_phase) = plt.subplots(2, 1, figsize=(12, 8), sharex=True)
colors = plt.cm.viridis(np.linspace(0, 1, len(resultados)))

for r, color in zip(resultados, colors):
    if r['freqs'] is not None and r['mag_db'] is not None:
        label_str = f"{PARAM_NAME} = {r['val']}"
        ax_mag.semilogx(r['freqs'], r['mag_db'], label=label_str, color=color, linewidth=1.8)
        ax_phase.semilogx(r['freqs'], r['phase_deg'], label=label_str, color=color, linewidth=1.8)

ax_mag.set_title(f'Respuesta en Frecuencia (Bode) - Barrido de {TARGET_COMPONENT} ({PARAM_NAME})', fontsize=13, fontweight='bold')
ax_mag.set_ylabel('Magnitud (dB)')
ax_mag.axhline(0, color='red', linestyle=':', alpha=0.6)
ax_mag.grid(True, which="both", alpha=0.3)
ax_mag.legend(loc='upper right', bbox_to_anchor=(1.15, 1.0))

ax_phase.set_xlabel('Frecuencia (Hz)')
ax_phase.set_ylabel('Fase (°)')
ax_phase.grid(True, which="both", alpha=0.3)

plt.tight_layout()
plt.show()

```
